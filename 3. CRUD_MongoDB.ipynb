# Ejemplo 1

```python
#  Importar la librería de PyMongo
from pymongo import MongoClient

#  Conectarse a MongoDB (por defecto al localhost en el puerto 27017)
client = MongoClient('mongodb+srv://jctovarg_db_user:UP2026@cluster0.jujmuoe.mongodb.net/')

# Eliminar la base de datos
# client.drop_database('universidad')

#  Seleccionar base de datos (se crea si no existe)
db = client['universidad']
```

```python
#  Seleccionar colección (tabla en SQL, se crea si no existe)
usuarios = db['estudiantes']

#  Limpiar colección para pruebas repetibles
usuarios.delete_many({})
```

##  CREATE: insertar documentos

```python
# Insertar un único documento
usuarios.insert_one({
    "nombre": "Ana",
    "edad": 25,
    "correo": "ana@email.com",
    "hobbies": ["cine", "música"]
})

# Insertar varios documentos a la vez
usuarios.insert_many([
    {"nombre": "Carlos", "edad": 30, "correo": "carlos@email.com", "hobbies": ["deporte"]},
    {"nombre": "Laura", "edad": 28, "correo": "laura@email.com", "hobbies": []}
])
```

##  READ: consultar documentos

```python
# Consultar todos los documentos de la colección
for usuario in usuarios.find():
    print(usuario)
print("*"*20)
# Consultar con condición (edad mayor a 26)
for usuario in usuarios.find({"edad": {"$gt": 26}}):
    print(usuario)
print("*"*20)
# Mostrar solo ciertos campos (proyección)
for usuario in usuarios.find({}, {"_id": 0, "nombre": 1, "edad": 1}):
    print(usuario)
```

##  UPDATE: actualizar documentos

```python
# Actualizar un único documento (cambiar edad de Ana a 26)
usuarios.update_one(
    {"nombre": "Ana"},
    {"$set": {"edad": 26}}
)

# Actualizar varios documentos (sumar +1 a todos los mayores de 26)
usuarios.update_many(
    {"edad": {"$gt": 26}},
    {"$inc": {"edad": 1}}
)

for usuario in usuarios.find({}, {"_id": 0, "nombre": 1, "edad": 1}):
    print(usuario)
```

##  UPSERT: actualizar o insertar si no existe

```python
# Si no existe un usuario llamado Pedro, lo crea
usuarios.update_one(
    {"nombre": "Pedro"},
    {"$set": {"edad": 23, "correo": "pedro@email.com"}},
    upsert=True
)

for usuario in usuarios.find({}, {"_id": 0, "nombre": 1, "edad": 1}):
    print(usuario)
```

##  DELETE: eliminar documentos

```python
# Eliminar un único documento (Laura)
usuarios.delete_one({"nombre": "Laura"})

# Eliminar todos los documentos con edad menor a 25
usuarios.delete_many({"edad": {"$lt": 25}})

for usuario in usuarios.find({}, {"_id": 0, "nombre": 1, "edad": 1, "hobbies": 1}):
    print(usuario)
```

# Ejemplo 2

```python
#  Importar la librería PyMongo
from pymongo import MongoClient

#  Conectarse a MongoDB (por defecto localhost)
client = MongoClient('mongodb+srv://jctovarg_db_user:UP2026@cluster0.jujmuoe.mongodb.net/')

#  Seleccionar base de datos
db = client['universidad']

#  Seleccionar colección
estudiantes = db['estudiantes_v2']

#  Limpiar colección para pruebas repetibles
estudiantes.delete_many({})
```

##  CREATE: insertar documentos con subdocumentos

```python
# Insertar un estudiante con documentos embebidos en 'direccion' y 'cursos'
estudiantes.insert_one({
    "nombre": "María",
    "edad": 22,
    "direccion": {
        "calle": "Av. Siempre Viva",
        "ciudad": "Lima",
        "codigo_postal": "15001"
    },
    "cursos": [
        {"nombre": "Matemáticas", "nota": 18},
        {"nombre": "Estadística", "nota": 17}
    ]
})

# Insertar varios estudiantes
estudiantes.insert_many([
    {
        "nombre": "Luis",
        "edad": 23,
        "direccion": {
            "calle": "Calle Falsa",
            "ciudad": "Arequipa",
            "codigo_postal": "04001"
        },
        "cursos": []
    },
    {
        "nombre": "Sara",
        "edad": 21,
        "direccion": {
            "calle": "Av. Central",
            "ciudad": "Cusco",
            "codigo_postal": "08002"
        },
        "cursos": [{"nombre": "Programación", "nota": 20}]
    }
])
```

##  READ: consultar documentos y subdocumentos

```python
# Consultar todos los estudiantes
for estudiante in estudiantes.find():
    print(estudiante)
print("*"*20)
# Consultar estudiantes de una ciudad específica (campo embebido)
for estudiante in estudiantes.find({"direccion.ciudad": "Lima"}):
    print(estudiante)
print("*"*20)
# Consultar cursos de un estudiante
for estudiante in estudiantes.find({"nombre": "María"}, {"_id": 0, "nombre":1, "cursos": 1}):
    print(estudiante)
```

##  UPDATE: actualizar documentos embebidos

```python
for estudiante in estudiantes.find({}, {"_id": 0, "nombre": 1, "direccion": 1, "cursos": 1}):
    print(estudiante)
```

```python
# Cambiar la ciudad de residencia de Luis
estudiantes.update_one(
    {"nombre": "Luis"},
    {"$set": {"direccion.ciudad": "Trujillo"}}
)

# Actualizar la nota de un curso específico en un array embebido
estudiantes.update_one(
    {"nombre": "María", "cursos.nombre": "Matemáticas"},
    {"$set": {"cursos.$.nota": 19}}
    #El símbolo $ representa el primer elemento del array cursos que cumplió la condición.
)

for estudiante in estudiantes.find({}, {"_id": 0, "nombre": 1, "direccion": 1, "cursos": 1}):
    print(estudiante)
```

```python
# Actualizar la nota de todos los cursos
estudiantes.update_one(
    {"nombre": "María"},
    {"$set": {"cursos.$[].nota": 20}}
    #$[] -> todos los elementos del array
)

for estudiante in estudiantes.find({}, {"_id": 0, "nombre": 1, "cursos": 1}):
    print(estudiante)
```

##  UPSERT: actualizar o insertar subdocumento si no existe

```python
# Si no existe un estudiante llamado Pedro, lo crea con subdocumento 'direccion'
estudiantes.update_one(
    {"nombre": "Pedro"},
    {"$set": {
        "edad": 20,
        "direccion": {"calle": "Av. Norte", "ciudad": "Piura", "codigo_postal": "20001"},
        "cursos": []
    }},
    upsert=True
)
for estudiante in estudiantes.find({}, {"_id": 0, "nombre": 1, "direccion": 1, "cursos": 1}):
    print(estudiante)
```

##  DELETE: eliminar documentos o elementos de subdocumentos

```python
# Eliminar a Sara de la colección
estudiantes.delete_one({"nombre": "Sara"})

for estudiante in estudiantes.find({}, {"_id": 0, "nombre": 1, "direccion": 1, "cursos": 1}):
    print(estudiante)
```
